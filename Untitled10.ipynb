```python
import pandas as pd
import numpy as np
df = pd.read_csv("C:/Users/ASUS/Downloads/crop_yield (1).csv")
```

```python
print("Size:", df.size)
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
print(df.isnull().sum())

df.fillna({
    "Production": df["Production"].median()
}, inplace=True)

df.drop_duplicates(inplace=True)

print(df.isnull().sum())
print("Shape:", df.shape)
```

```python
df["Crop_Year"] = df["Crop_Year"].astype(int)

print(df.dtypes)
```

```python
print(df.loc[0:4, ["State_Name", "Crop", "Production"]])

print(df.iloc[0:5, 0:4])
```

```python
print(df[df["Area"] > 1000])

print(df[df["Production"] > 100])

print(df[
    (df["Area"] > 1000) &
    (df["Production"] > 100)
])
```

```python
df["Production_per_Area"] = (
    df["Production"] / df["Area"].replace(0, np.nan)
)

print(df.head())
```

```python
print(df.groupby("Crop")["Production"].sum())

print(df.groupby("Season")["Production"].mean())

print(df.groupby("State_Name")["Area"].sum())
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["Crop_Encoded"] = le.fit_transform(df["Crop"])

print(df[["Crop", "Crop_Encoded"]].head())
```

```python
data = df["Production"]

print("Mean:", data.mean())
print("Median:", data.median())
print("Mode:", data.mode().tolist())

print("Standard Deviation:", data.std())
print("Variance:", data.var())

print("Minimum:", data.min())
print("Maximum:", data.max())
print("Range:", data.max() - data.min())

Q1 = data.quantile(0.25)
Q2 = data.quantile(0.50)
Q3 = data.quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)

print("Skewness:", data.skew())
print("Kurtosis:", data.kurt())

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[(data < lower) | (data > upper)]

print("Lower limit:", lower)
print("Upper limit:", upper)
print("Number of outliers:", len(outliers))

```
